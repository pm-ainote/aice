## 문제 1. [결과 예측]
아래 코드를 실행하면 무엇이 나올까요? 나오는 글자를 따옴표로 감싸 `q1` 에 저장하세요.
```python
x = 5
if x >= 10:
    print('큼')
else:
    print('작음')
```

````python
q1 = '작음'   # 5 는 10 이상이 아니라 else 쪽
````

**해설** 5 는 10 이상이 아니라서 조건이 False 가 되고, else 아래의 코드가 실행돼요. 그래서 '작음' 이에요.

## 문제 2. [결과 예측]
아래 코드를 실행하면 마지막에 나오는 숫자는 무엇일까요? 그 숫자를 `q2` 에 저장하세요.
```python
for i in range(4):
    print(i)
```

````python
q2 = 3   # 0, 1, 2, 3 - 괄호 안의 4 는 포함 안 됨
````

**해설** range 괄호 안에 4 를 넣으면 0 부터 숫자 네 개, 즉 0, 1, 2, 3 을 만들어요. 괄호 안의 4 는 포함하지 않으니 마지막은 3 이에요.

## 문제 3. [오류 수정]
아래 셀은 실행하면 SyntaxError 가 나요. 한 군데를 고쳐서 `result` 에 합격이 들어가게 하세요.

````python
score = 85
if score >= 80:   # if 줄 끝에 콜론
    result = '합격'
````

**해설** if 줄 끝에 콜론이 없으면 SyntaxError 가 나고, 마지막 줄에 expected ':' 가 보여요. 콜론은 "여기까지가 조건" 이라는 표시예요.

## 문제 4. [빈칸 채우기]
scores 의 점수를 하나씩 꺼내 합계를 구하는 코드예요. 빈칸(____)을 채우세요.

````python
scores = [80, 90, 70]
total = 0
for s in scores:
    total = total + s
q4 = total
````

**해설** for 이름 in 리스트: 모양이에요. in 은 "리스트 안에 있는 값" 이라는 뜻이고, 줄 끝에는 콜론을 붙여요.

## 문제 5. [직접 작성]
변수 `age` 에 15 를 넣으세요. 그리고 age 가 20 이상이면 `answer` 에 `'성인'` 을, 아니면 `'미성년'` 을 저장하는 조건문을 작성하세요. 마지막에 print 로 answer 를 확인하세요.

````python
age = 15
if age >= 20:
    answer = '성인'
else:
    answer = '미성년'
print(answer)
````

**해설** 15 는 20 이상이 아니라서 else 아래의 코드가 실행돼요. if 와 else 는 같은 위치에서 시작하고, 각각 실행할 코드는 한 번 들여써요.
