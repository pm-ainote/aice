## 문제 1. [결과 예측]
`nums = [10, 20, 30]` 일 때 `nums[2]` 의 값을 `q1` 에 저장하세요.

```python
q1 = 30   # 0번부터 세면 2번은 세 번째
```

**해설** 자리 번호는 0부터 세니까 2번은 세 번째 값인 30이에요.

## 문제 2. [결과 예측]
`info = {'name': '피노', 'age': 40}` 일 때 `info['score']` 를 실행하면 나오는 에러 이름을 따옴표로 감싸 `q2` 에 저장하세요.

```python
q2 = 'KeyError'   # score 라는 키가 없어요
```

**해설** info 에는 'name' 과 'age' 만 있어서, 없는 딕셔너리 키 'score' 를 꺼내면 KeyError 가 나요.

## 문제 3. [빈칸 채우기]
리스트에 값이 몇 개 들어 있는지 세는 함수 이름으로 빈칸(____)을 채우세요.

```python
fruits = ['사과', '배', '귤']
q3 = len(fruits)
```

**해설** 리스트에 값이 몇 개 들어 있는지는 len 으로 확인해요. len 은 길이를 뜻하는 length 를 줄인 이름이에요.

## 문제 4. [오류 수정]
마지막 값(노랑)을 꺼내려던 코드인데 IndexError 가 나요. 자리 번호를 고쳐 실행하세요.

```python
colors = ['빨강', '파랑', '노랑']
q4 = colors[2]   # 값이 세 개면 마지막 자리 번호는 2
```

**해설** 값이 세 개면 자리 번호는 0, 1, 2 예요. 3번은 없어서 IndexError 가 나요. 마지막 값은 2번이에요.

## 문제 5. [직접 작성]
딕셔너리 `pet` 을 만드세요. 딕셔너리 키 `'name'` 의 값은 `'초코'`, 딕셔너리 키 `'age'` 의 값은 숫자 `3` 이에요. 그리고 `pet` 에서 `'name'` 의 값을 꺼내 `answer` 에 저장하세요.

```python
pet = {'name': '초코', 'age': 3}
answer = pet['name']
```

**해설** 딕셔너리는 중괄호로 감싸고 '키': 값 짝을 쉼표로 구분해요. 3 은 숫자라 따옴표 없이 써요. 값은 딕셔너리 이름 뒤 대괄호에 키를 넣어 꺼내요.
