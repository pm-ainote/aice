## 4. [출력 예측] value_counts = 많은 순서로

## 5. [빈칸] normalize=True → 비율

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (normalize=True → 비율):
#   df['Sex'].value_counts(normalize=True)
round(df['Sex'].value_counts(normalize=True)['male'], 3)
```

## 6. [오류 정정] 평균 = mean, 중앙값 = median

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (평균 = mean, 중앙값 = median):
#   df.groupby('Pclass')['Fare'].mean()
df.groupby('Pclass')['Fare'].mean()
```

## 7. [빈칸] 그리고 = &, 또는 = |

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (그리고 = &, 또는 = |):
#   df[(df['Pclass'] == 1) & (df['Fare'] >= 100)]
len(df[(df['Pclass'] == 1) & (df['Fare'] >= 100)])
```

## 8. [오류 정정] 칸 제목 = axes[i].set_title

## 9. [빈칸] hue = 색으로 나누기

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (hue = 색으로 나누기):
#   sns.histplot(data=df, x='Age', hue='Survived')
import seaborn as sns
ax = sns.histplot(data=df, x='Age', hue='Survived')
ax.get_legend() is not None
```

## 11. [빈칸] (추가) 가장 많은 값의 이름 = idxmax

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (가장 많은 값의 이름 = idxmax):
#   df['Embarked'].value_counts().idxmax()
df['Embarked'].value_counts().idxmax()
```

## 12. [출력 예측] (추가) 1등석 요금 중앙값이 전체보다 크다

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (1등석 요금 중앙값이 전체보다 크다):
#   True
bool(df[df['Pclass'] == 1]['Fare'].median() > df['Fare'].median())
```

## 13. [오류 정정] (추가) head 는 함수: 괄호

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (head 는 함수: 괄호):
#   df.sort_values('Fare', ascending=False).head(3)
df.sort_values('Fare', ascending=False).head(3)
```

## 14. [빈칸] (추가) 두 컬럼 개수 표 = crosstab

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (두 컬럼 개수 표 = crosstab):
#   pd.crosstab(df['Sex'], df['Survived'])
int(pd.crosstab(df['Sex'], df['Survived']).loc['female', 1])
```
