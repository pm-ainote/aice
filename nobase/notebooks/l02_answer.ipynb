## 문제 1. [결과 예측]

```python
q1 = 4   # 1 에서 시작해 1 을 세 번 더해요
```

**해설** x 는 1 에서 시작해 `x = x + 1` 을 실행할 때마다 1 씩 커져요. 세 번 실행했으니 1 → 2 → 3 → 4 예요. (2강 「같은 셀, 다른 결과」)

## 문제 2. [오류 수정]

```python
y = 5
del y
# 같은 줄이 두 번 있으면 이미 지운 y 를 또 지워 NameError (한 줄을 지움)
q2 = 1
```
